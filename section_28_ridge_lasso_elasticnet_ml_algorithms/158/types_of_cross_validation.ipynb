### types of cross validation

**Cross-validation (CV)** is a resampling technique used to evaluate how well a machine learning model will generalize to unseen, real-world data.

Instead of just splitting your data once into a training set and a testing set, CV splits the data multiple times. This ensures your model's performance score isn't just a lucky (or unlucky) fluke based on one random split.

Here is a quick breakdown of how each method works and when to use it:

### 1. Leave-One-Out CV (LOOCV)

You use a single data point as the validation set, and the remaining $n-1$ points as the training set. You repeat this process $n$ times so that every single data point gets to act as the test set exactly once.

* **Pros:** Highly unbiased because you train on almost all the data every time.
* **Cons:** Extremely computationally expensive for large datasets, as you have to train the model from scratch $n$ times.

### 2. Leave-$p$-Out CV (LpOCV)

Instead of leaving one observation out, you leave $p$ observations out for the validation set, and train on the rest. You repeat this for **every possible combination** of $p$ items in your dataset.

* **Pros:** Thoroughly tests the model on all possible subsets.
* **Cons:** Computationally brutal. Even for relatively small datasets, the number of combinations quickly becomes mathematically explosive. It is rarely used in practice.

### 3. K-Fold CV

The dataset is randomly shuffled and divided into $k$ equal-sized groups (called "folds"). The model is trained on $k-1$ folds and tested on the 1 remaining fold. This is repeated $k$ times so that every fold gets to be the test set exactly once. The final performance is the average of the $k$ tests.

* **Pros:** The industry standard (usually $k=5$ or $k=10$). It strikes the best balance between computational cost and generating a reliable, low-variance performance estimate.

### 4. Stratified K-Fold CV

This works exactly like standard K-Fold CV, but with one critical rule: it enforces that the **ratio of target classes in each fold matches the ratio in the entire dataset**.

* **When to use it:** It is absolutely essential for imbalanced classification problems. If your dataset is 99% healthy patients and 1% sick patients, a random standard K-Fold might create a fold with zero sick patients. Stratified K-Fold ensures every single fold contains exactly 1% sick patients.

### 5. Time Series CV

You cannot randomly shuffle time-series data because chronological order matters — you cannot use future data to predict the past (this is called data leakage). Instead, Time Series CV uses an **expanding window** (or rolling window) approach. You train on a chronological sequence of data (e.g., days 1-3) to predict the immediate future (day 4), then expand your training set (days 1-4) to predict the next step (day 5), and so on.
