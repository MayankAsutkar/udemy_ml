## algerian forest fires dataset



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header = 1)
```

```python
df.head()
```

```python
df.info()
```

```python
## data cleaning process
df.isnull().sum()
```

```python
df.shape
```

```python
df[df.isnull().any(axis = 1)]
```

```python
df.loc[:122,"Region"] = 0
df.loc[122:,"Region"] = 1
```

```python
df.info()
```

```python
df[['Region']] = df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis = 1)]
```

```python
## removing null values
df = df.dropna().reset_index(drop = True)
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df = df.drop(122).reset_index(drop =True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## fix spaces in column names
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
df.head()
```

```python
## change the required columns as integer data type
df[['month','day','year','Temperature','RH','Ws']]  = df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
### changing the other columns to float data types
s = [features for features in df.columns if df[features].dtypes=='str']
```

```python
for i in s:
    if i != 'Classes':
        df[i] = df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.to_csv("cleaned_algerian_forest_fires.csv", index=False)
```

### eda and feature engineering

```python
df_copy = df.copy()
```

```python
df_copy.head()
```

```python
df_copy.drop(['day','month','year'],axis = 1,inplace=True)
```

```python
## encoding of categories in classes
df_copy['Classes'] = np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df['Classes'] = np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
### plot density plot for all features
plt.style.use('seaborn-v0_8')
df_copy.hist(bins = 50,figsize = (20,15))
plt.show()
```

```python
## percentage of pie chart
percentage = df_copy['Classes'].value_counts(normalize=True)*100
print(percentage)
```

```python
## plotting pie chart
classlabels = ['Fire','Not fire']
plt.figure(figsize = (12,7))
plt.pie(percentage,labels = classlabels,autopct='%1.1f%%')
plt.title("Pie chart of classes")
```
