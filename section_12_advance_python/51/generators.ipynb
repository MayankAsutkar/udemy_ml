# generators

generators are subclass of iterators itself

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(3):
        return i**2
```

```python
square(3)
```

```python
def square(n):
    for i in range(3):
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
```

```python
try:
    print(next(a))
except StopIteration:
    print("access outside the limit")
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen = my_generator()
```

```python
try:
    print(next(gen))
except StopIteration:
    print("access outside the limit")
```

```python
for val in gen:
    print(val)
```

```python
### practical example: reading large files

def read_large_files(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt'

for line in read_large_files(file_path):
    print(line.strip())
```
