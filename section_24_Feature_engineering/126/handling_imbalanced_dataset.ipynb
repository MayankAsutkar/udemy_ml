## handling imbalanced dataset : 
-> if the dataset is not distributed properly the model prediction result maybe biased 

-> so to maintain the fairness we fix the dataset :
1. upsampling
2. downsampling



```python
import numpy as np
import pandas as pd

# set the random seed for reproducibility
np.random.seed(123)

# create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

```

```python
n_class_0,n_class_1
```

```python
## CREATE MY DATAFRAME WITH IMBALANCED DATASET
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc = 0, scale = 1, size = n_class_0),
    'feature_2': np.random.normal(loc = 0, scale = 1, size = n_class_0),
    'target' : [0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc = 0, scale = 1, size = n_class_1),
    'feature_2': np.random.normal(loc = 0, scale = 1, size = n_class_1),
    'target' : [1]*n_class_1
})
```

```python
class_0
```

```python
class_1
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop = True)
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## upsampling
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample
df_minority_upsampled = resample(
            df_minority,replace=True, ## ssample with replacement
            n_samples = len(df_majority),
            random_state=42
         )
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```

```python
## downsampling

import numpy as np
import pandas as pd

# set the random seed for reproducibility
np.random.seed(123)

# create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

## CREATE MY DATAFRAME WITH IMBALANCED DATASET
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc = 0, scale = 1, size = n_class_0),
    'feature_2': np.random.normal(loc = 0, scale = 1, size = n_class_0),
    'target' : [0]*n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc = 0, scale = 1, size = n_class_1),
    'feature_2': np.random.normal(loc = 0, scale = 1, size = n_class_1),
    'target' : [1]*n_class_1
})
df = pd.concat([class_0,class_1]).reset_index(drop = True)

## check the class distribution

print(df['target'].value_counts())
```

```python
## downsampling
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]

from sklearn.utils import resample
df_majority_downsampled = resample(
            df_majority,replace=False, 
            n_samples = len(df_minority),
            random_state=42
         )

df_downsampled = pd.concat([df_minority,df_majority_downsampled])
```
