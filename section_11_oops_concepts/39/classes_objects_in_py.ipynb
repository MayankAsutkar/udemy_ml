## classes and objects

Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modeling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including instance variables and methods.

```python
### a class is a blue print for creating objects. Attibutes, methods

class Car:
    pass

audi = Car()
bmw = Car()

# here  audi and bmw are objects

print(type(audi))
```

```python
print(audi)
print(bmw)

## output will be in form of class name and sprcific memory location
```

```python
tata = Car()
tata.windows = 4
```

```python
dir(tata)
## this will give different methods and attributes of tata object
```

```python
### instance variable and methods

class Dog:
    ## constructor
    def __init__(self,name,age): ## self : To differentiate between local variables and object attributes
        self.name = name
        self.age = age
        
## create objects
dog1 = Dog("buddy",3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2 = Dog("lucy",4)
print(dog2)
print(dog2.name)
print(dog2.age)
```

```python
### defie a class with instance methods

class Dog:
    def __init__(self, name,age):
        self.name = name
        self.age = age
    def bark(self):
        print(f"{self.name} says woof")
        
dog1 = Dog("tom",2)
dog1.bark()
dog2 = Dog("lucifer",4)
dog2.bark()
```

```python
### Modeling a bank account

## define a class for bank account
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
        
    def deposit(self,amount):
        self.balance += amount
        print(f"{amount} is deposited and new balance is:{self.balance} ")
    def withdraw(self,amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")
    def get_balance(self):
        return self.balance
    
## create an account

account = BankAccount("mayank",27903)
account.get_balance()
account.deposit(200)
account.withdraw(2301)
```
