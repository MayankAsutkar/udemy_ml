## inheritance in python

Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance.

```python
## Inheritance
## parent class
## single inheritance
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
        
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
        return
    
```

```python
car1 = Car(4,5,"petrol")
car1.drive()
```

```python
class Tesla(Car): ## this is how we inherit
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype) ## we  are calling init method from parent class
        self.is_selfdriving = is_selfdriving
        
    def selfdriving(self):
        print(f"Tesla supports seflf driving {self.is_selfdriving}")
        
```

```python
tesla = Tesla(4,5,"Electric",True)
tesla.selfdriving()
tesla.drive()
```

```python
### multiple inheritance
## when a class  inherits more than oone base class.

## Base class 1
class Animal:
    def __init__(self,name):
        self.name = name
    
    def speak(self):
        print("Subclass must implement this method")
        
## base class 2
class Pet:
    def __init__(self,owner):
        self.owner = owner
        
## derived class
class Dog(Animal,Pet): ## this  is how you do multiple inheritance
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
        
    def speak(self):
        return f"{self.name} says woof"

    
```

```python
## creating an object
dog = Dog("buddy","mayank")
print(dog.speak())
print(f"Owner : {dog.owner}")
```
