### Abstraction

Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexity and

```python
from abc import ABC,abstractmethod

## Abstract class
class Vehicle(ABC):
    def drive(self):
        print("the vehilcle is used for driving")
        
    @abstractmethod
    def start_engine(self):
        pass
    
class Car(Vehicle):
    def start_engine(self):
        return "The car engine is started"
    
def operate_vehicle(vehicle):
    vehicle.drive()
    return vehicle.start_engine()
    
car = Car()
print(operate_vehicle(car))
```
